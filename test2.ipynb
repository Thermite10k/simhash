```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName("test2").master("local[*]").getOrCreate()
```

```python
data = list(range(20))
myrdd = spark.sparkContext.parallelize(data, numSlices=4)
```

```python
myrdd
```

# Transformations

Transformations return a new RDD

Lazy evaluation

Collect is an action and return

```python
def sq(x): return x*x

squared = myrdd.map(lambda x: (x*x, 3*x))
squared_flat = myrdd.flatMap(lambda x: (x*x, 3*x))
print(squared.collect())
print(squared_flat.collect())
```

#Filter

Select a subset of the data.

```python
even = myrdd.filter(lambda x: x%2 == 0)
even.collect()
```

# Sample

Draw a random sample of the data, with or without replacement, let's go with 20%

```python
myrdd.sample(withReplacement=False, fraction=0.20).collect()
```

# Pseudo Set operations

```python
myrdd2 = spark.sparkContext.parallelize(range(10, 30))
union = myrdd.union(myrdd2)
```

```python
print(union.collect())
```

```python
intersection = myrdd.intersection(myrdd2)
print(intersection.collect())
```
